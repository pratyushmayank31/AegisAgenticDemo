```python
### CASE PREPARATION — INITIALIZATION

from pathlib import Path
from datetime import datetime, timezone
import hashlib
import json
import os
import sqlite3
import sys
import uuid

EXPECTED_PROJECT_ROOT = Path(
    "/Users/pmayank/workspace/LdcDemo"
).resolve()

SAMPLE_INVOICE_DIRECTORY = (
    EXPECTED_PROJECT_ROOT / "sample_invoices"
)

DATABASE_PATH = (
    EXPECTED_PROJECT_ROOT / "data" / "finance_demo.db"
)

SESSION_PATH = (
    EXPECTED_PROJECT_ROOT / "data" / "demo_session.json"
)

# Ensure consistent relative paths and project imports.
os.chdir(EXPECTED_PROJECT_ROOT)

if str(EXPECTED_PROJECT_ROOT) not in sys.path:
    sys.path.insert(0, str(EXPECTED_PROJECT_ROOT))

initialization_checks = {
    "Project root": EXPECTED_PROJECT_ROOT.exists(),
    "Sample invoice directory": SAMPLE_INVOICE_DIRECTORY.exists(),
    "Finance database": DATABASE_PATH.exists(),
    "Data directory": SESSION_PATH.parent.exists(),
}

print("=" * 72)
print("CASE PREPARATION — INITIALIZATION")
print("=" * 72)
print(f"Project root      : {EXPECTED_PROJECT_ROOT}")
print(f"Invoice directory : {SAMPLE_INVOICE_DIRECTORY}")
print(f"Finance database  : {DATABASE_PATH}")
print(f"Session file      : {SESSION_PATH}")
print("-" * 72)

for check_name, passed in initialization_checks.items():
    symbol = "✅" if passed else "❌"
    status = "READY" if passed else "MISSING"
    print(f"{symbol} {check_name:<28} {status}")

case_preparation_ready = all(initialization_checks.values())

print("-" * 72)

if case_preparation_ready:
    print("✅ CASE PREPARATION ENVIRONMENT READY")
else:
    print("❌ CASE PREPARATION ENVIRONMENT NOT READY")

assert case_preparation_ready, (
    "Resolve the missing case-preparation dependencies before continuing."
)
```
